### Slide 3: Problem Formulation
* **Target Type:** Supervised Continuous Regression ($y \in [0.0, 100.0]$).
* **Engineering Rationale:** Avoids loss of granularity inherent to binary classification and permits dynamic institutional decision thresholds.

```python
# ==============================================================================
# 2. DATA INGESTION & VARIABLE PROFILING
# ==============================================================================
import pandas as pd
import numpy as np

df = pd.read_csv('data/StudentPerformanceFactors.csv')
print(f"[*] Dataset Loaded: {df.shape[0]} rows, {df.shape[1]} columns.")

TARGET = 'Exam_Score'
X = df.drop(columns=[TARGET])
y = df[TARGET]

num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X.select_dtypes(exclude=[np.number]).columns.tolist()

print(f"[*] Quantitative Features ({len(num_cols)}): {num_cols}")
print(f"[*] Qualitative Features  ({len(cat_cols)}): {cat_cols}")
df.head()
```

```python
# ==============================================================================
# 3. TRAIN/TEST SPLIT & COLUMN TRANSFORMER PIPELINE
# ==============================================================================
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

num_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer([
    ('num', num_transformer, num_cols),
    ('cat', cat_transformer, cat_cols)
])

print(f"Training set: {X_train.shape[0]} | Holdout test set: {X_test.shape[0]}")
```

```python
# ==============================================================================
# 4. 5-FOLD CROSS-VALIDATION BENCHMARKING (5 CANDIDATE ALGORITHMS)
# ==============================================================================
from sklearn.dummy import DummyRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.model_selection import KFold, cross_validate

models = {
    "Dummy Baseline": DummyRegressor(strategy='mean'),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42),
    "Linear Regression (Selected)": LinearRegression()
}

kf = KFold(n_splits=5, shuffle=True, random_state=42)
metrics = ['r2', 'neg_mean_absolute_error', 'neg_root_mean_squared_error']

print(f"{'Model':<30} | {'CV R2':<8} | {'CV MAE':<8} | {'CV RMSE':<8}")
print("-" * 62)

for name, estimator in models.items():
    pipe = Pipeline([('preprocessor', preprocessor), ('regressor', estimator)])
    res = cross_validate(pipe, X_train, y_train, cv=kf, scoring=metrics, n_jobs=-1)
    print(f"{name:<30} | {res['test_r2'].mean():<8.4f} | {-res['test_neg_mean_absolute_error'].mean():<8.4f} | {-res['test_neg_root_mean_squared_error'].mean():<8.4f}")
```

```python
# ==============================================================================
# 5. FINAL PRODUCTION TRAINING, HOLDOUT EVALUATION & SERIALIZATION
# ==============================================================================
import joblib
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.inspection import permutation_importance

final_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])

final_pipeline.fit(X_train, y_train)
y_pred = final_pipeline.predict(X_test)
residuals = y_test - y_pred

test_r2 = r2_score(y_test, y_pred)
test_mae = mean_absolute_error(y_test, y_pred)
test_rmse = np.sqrt(mean_squared_error(y_test, y_pred))
res_mean = residuals.mean()
res_std = residuals.std()

print(f"Holdout Test R2:            {test_r2:.4f}")
print(f"Holdout Test MAE:           {test_mae:.4f}")
print(f"Holdout Test RMSE:          {test_rmse:.4f}")
print(f"Residual Mean (Zero-bias):  {res_mean:.4f}")
print(f"Residual Std Dev:           {res_std:.4f}")

# Permutation Feature Importance on Test Partition
perm_res = permutation_importance(final_pipeline, X_test, y_test, n_repeats=10, random_state=42, n_jobs=-1)
importance_df = pd.DataFrame({
    'Feature': X_test.columns,
    'Importance_Mean': perm_res.importances_mean,
    'Importance_Std': perm_res.importances_std
}).sort_values(by='Importance_Mean', ascending=False)

os.makedirs('models', exist_ok=True)
joblib.dump(final_pipeline, 'models/exam_score_prediction_pipeline.joblib')
print("\n[✓] Pipeline serialized to models/exam_score_prediction_pipeline.joblib")
```
